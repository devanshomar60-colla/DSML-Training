```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sudalairajkumar/covid19-in-india")

print("Path to dataset files:", path)
```

```python
import os
print(os.listdir(path))
```

```python
import pandas as pd
file_path= os.path.join(path,"covid_19_india.csv")
df=pd.read_csv(file_path)
df.tail(2)
```

```python
#undersatnd the dataset
print("Shape", df.shape)
# Dataset ka shape batata hai
# Pehla value = total rows
# Dusra value = total columns
print("Data type", df.dtypes)
print("describe",df.describe)
# Numerical columns ka statistical summary deta hai
# Isme count, mean, standard deviation, minimum,
# 25%, 50%, 75% percentile aur maximum values hoti hain
```

```python
df.head(10)
# checking the upper 10 rows to understand the dataset
```

```python

import matplotlib.pyplot as plt
state_data = df.groupby("State/UnionTerritory")["Confirmed"].max().reset_index()

top_10 = state_data.sort_values("Confirmed", ascending=False).head(10)

print(top_10)
plt.bar(top_10["State/UnionTerritory"], top_10["Confirmed"])

plt.xlabel("State/UnionTerritory")
plt.ylabel("Confirmed Cases")
plt.title("Top 10 States According to Confirmed Cases")

plt.xticks(rotation=90)

plt.show()
```

```python
import matplotlib.pyplot as plt
state_data=df.groupby("State/UnionTerritory")["Deaths"].max().reset_index()
top_10=state_data.sort_values("Deaths",ascending=False).head(10)

plt.bar(top_10['State/UnionTerritory'],top_10["Deaths"])
plt.xlabel("State/UnionTerritory")
plt.ylabel("Deaths")
plt.title("Top 10 States According to Cured")

plt.xticks(rotation=90)

plt.show()
```

```python
import matplotlib.pyplot as plt
state_data=df.groupby("State/UnionTerritory")["Cured"].max().reset_index()
top_10=state_data.sort_values("Cured",ascending=False).head(10)

plt.bar(top_10["State/UnionTerritory"], top_10["Cured"], color=[ "cyan", "gray"])

plt.xlabel("State/UnionTerritory")
plt.ylabel("Cured")
plt.title("Top 10 States According to Cured")

plt.xticks(rotation=90)

plt.show()

```
